```python
import itertools
import json
import os

import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np

OUT = "ketqua"
os.makedirs(OUT, exist_ok=True)
```

```python
def trimf(x, a, b, c):
    x = np.asarray(x, dtype=float)
    left = np.where(b > a, (x - a) / (b - a + 1e-12), 1.0)
    right = np.where(c > b, (c - x) / (c - b + 1e-12), 1.0)
    return np.clip(np.minimum(left, right), 0.0, 1.0)
```

```python
def trapmf(x, a, b, c, d):
    x = np.asarray(x, dtype=float)
    left = np.where(b > a, (x - a) / (b - a + 1e-12), 1.0)
    right = np.where(d > c, (d - x) / (d - c + 1e-12), 1.0)
    return np.clip(np.minimum(np.minimum(left, 1.0), right), 0.0, 1.0)
```

```python
# Mỗi biến: (miền giá trị, {nhãn: (hàm, tham số)}); thứ tự nhãn = mức nguy hiểm tăng dần (0,1,2)
VARS_IN = {
    "behav": {
        "name": "Số hành vi bất thường",
        "unit": "hành vi",
        "universe": (0, 30),
        "labels": ["Ít", "Vừa", "Nhiều"],
        "mf": [(trapmf, (0, 0, 3, 10)), (trimf, (5, 12, 19)), (trapmf, (14, 21, 30, 30))],
    },
    "sig": {
        "name": "Chữ ký số",
        "unit": "điểm hợp lệ (0=không hợp lệ, 10=hợp lệ)",
        "universe": (0, 10),
        # nhãn xếp theo mức nguy hiểm tăng dần: Hợp lệ(0), Không có(1), Không hợp lệ(2)
        "labels": ["Hợp lệ", "Không có", "Không hợp lệ"],
        "mf": [(trapmf, (6.5, 8.5, 10, 10)), (trimf, (2.5, 5, 7.5)), (trapmf, (0, 0, 1.5, 3.5))],
    },
    "av": {
        "name": "Phản hồi Antivirus",
        "unit": "% engine cảnh báo",
        "universe": (0, 100),
        "labels": ["An toàn", "Ít cảnh báo", "Nhiều cảnh báo"],
        "mf": [(trapmf, (0, 0, 2, 8)), (trimf, (4, 18, 40)), (trapmf, (25, 45, 100, 100))],
    },
}
Y_UNIVERSE = np.linspace(0, 100, 1001)
Y_LABELS = ["An toàn", "Nghi vấn", "Rất nguy hiểm"]
Y_MF = [
    trapmf(Y_UNIVERSE, 0, 0, 15, 35),
    trimf(Y_UNIVERSE, 25, 50, 75),
    trapmf(Y_UNIVERSE, 65, 85, 100, 100),
]


def fuzzify(key, value):
    v = VARS_IN[key]
    return np.array([f(value, *p) for f, p in v["mf"]]).reshape(-1)
```

```python

# 2. Cơ sở luật: 27 luật đầy đủ (3 x 3 x 3)
# ----------------------------------------------------------------------------
W_BEHAV, W_SIG, W_AV = 1.5, 1.25, 2.0   # AV được tin cậy nhất, rồi hành vi, rồi chữ ký


def rule_conclusion(b, s, a):
    """b, s, a in {0,1,2} là mức nguy hiểm của từng đầu vào. Trả về chỉ số nhãn đầu ra."""
    score = W_BEHAV * b + W_SIG * s + W_AV * a
    if score < 2.5:
        return 0   # An toàn
    if score > 6.0:
        return 2   # Rất nguy hiểm
    return 1       # Nghi vấn


RULES = [(b, s, a, rule_conclusion(b, s, a)) for b, s, a in itertools.product(range(3), range(3), range(3))]
```

```python
# 3. Suy diễn Mamdani: AND = min, hợp thành = min, gộp = max, giải mờ = trọng tâm
# ----------------------------------------------------------------------------
def infer(behav, sig_score, av_pct, detail=False, tnorm="min", defuzz="centroid"):
    """tnorm: 'min' (mặc định) hoặc 'prod' cho phép VÀ; defuzz: 'centroid' hoặc 'mom'."""
    mu_b = fuzzify("behav", behav)
    mu_s = fuzzify("sig", sig_score)
    mu_a = fuzzify("av", av_pct)
    agg = np.zeros_like(Y_UNIVERSE)
    fired = []
    for i, (b, s, a, c) in enumerate(RULES):
        if tnorm == "min":
            w = min(mu_b[b], mu_s[s], mu_a[a])
        else:
            w = mu_b[b] * mu_s[s] * mu_a[a]
        if w > 0:
            agg = np.maximum(agg, np.minimum(w, Y_MF[c]))
            fired.append((i + 1, w, c))
    if agg.sum() < 1e-9:
        y = 50.0
    elif defuzz == "centroid":
        y = float((Y_UNIVERSE * agg).sum() / agg.sum())
    else:  # trung bình các điểm đạt cực đại (mean of maxima)
        y = float(Y_UNIVERSE[agg >= agg.max() - 1e-9].mean())
    if detail:
        return y, (mu_b, mu_s, mu_a), fired, agg
    return y


def classify(y):
    """Gán nhãn theo hàm thuộc đầu ra lớn nhất tại điểm y."""
    mus = [float(np.interp(y, Y_UNIVERSE, m)) for m in Y_MF]
    return Y_LABELS[int(np.argmax(mus))]


def dominant_label_index(key, value):
    return int(np.argmax(fuzzify(key, value)))


def reference_label(behav, sig_score, av_pct):
    """Nhãn tham chiếu (rõ): lấy nhãn trội nhất của từng đầu vào rồi áp dụng bảng luật."""
    b = dominant_label_index("behav", behav)
    s = dominant_label_index("sig", sig_score)
    a = dominant_label_index("av", av_pct)
    return Y_LABELS[rule_conclusion(b, s, a)]
```

```python
# 4. Thực nghiệm
# ----------------------------------------------------------------------------
SCENARIOS = [
    ("Phần mềm văn phòng chính hãng",          1, 10.0, 0),
    ("Công cụ nội bộ chưa ký, sạch AV",        2, 5.0, 0),
    ("Bản cài đặt đã ký, 1 engine báo nhầm",   3, 9.0, 3),
    ("Tiện ích không ký, hành vi vừa",         12, 5.0, 3),
    ("Trình cài đặt có chữ ký hết hạn",        8, 1.0, 5),
    ("File không ký, 12% engine cảnh báo",     10, 5.0, 12),
    ("Đã ký hợp lệ nhưng nhiều engine báo",    6, 9.5, 50),
    ("Crack/keygen không ký, AV báo cao",      15, 5.0, 55),
    ("Chữ ký giả mạo, hành vi vừa, AV ít",     14, 0.5, 15),
    ("Dropper chữ ký sai, hành vi nhiều",      22, 0.5, 35),
    ("Ransomware mẫu, mọi chỉ số cao",         28, 0.0, 85),
    ("Trojan không ký, hành vi nhiều",         25, 5.0, 60),
]


def run_scenarios():
    rows = []
    for name, b, s, a in SCENARIOS:
        y, mus, fired, agg = infer(b, s, a, detail=True)
        rows.append({
            "name": name, "behav": b, "sig": s, "av": a,
            "y": round(y, 2), "pred": classify(y), "ref": reference_label(b, s, a),
            "n_fired": len(fired),
        })
    return rows


def run_random_validation(n=5000, seed=2026):
    rng = np.random.default_rng(seed)
    bs = rng.uniform(0, 30, n)
    ss = rng.uniform(0, 10, n)
    as_ = rng.uniform(0, 100, n)
    ys = np.array([infer(*t) for t in zip(bs, ss, as_)])
    pred = np.array([classify(y) for y in ys])
    ref = np.array([reference_label(*t) for t in zip(bs, ss, as_)])
    agree = float((pred == ref).mean())
    conf = {r: {p: int(((ref == r) & (pred == p)).sum()) for p in Y_LABELS} for r in Y_LABELS}
    # sai lệch chỉ lệch 1 mức?
    idx = {l: i for i, l in enumerate(Y_LABELS)}
    off2 = int(sum(abs(idx[p] - idx[r]) == 2 for p, r in zip(pred, ref)))
    return agree, conf, off2, n, ys


def run_corner_validation():
    """27 tổ hợp 'điển hình' (tâm của từng nhãn) so với kết luận của bảng luật."""
    centers = {
        "behav": [1, 12, 25],
        "sig": [9.5, 5, 0.5],
        "av": [0, 18, 70],
    }
    rows = []
    ok = 0
    for b, s, a in itertools.product(range(3), range(3), range(3)):
        y = infer(centers["behav"][b], centers["sig"][s], centers["av"][a])
        exp = Y_LABELS[rule_conclusion(b, s, a)]
        got = classify(y)
        ok += exp == got
        rows.append((VARS_IN["behav"]["labels"][b], VARS_IN["sig"]["labels"][s],
                     VARS_IN["av"]["labels"][a], exp, round(y, 1), got))
    return rows, ok
```

```python
# 6. Thực nghiệm bổ sung
# ----------------------------------------------------------------------------
def _samples(n, seed):
    rng = np.random.default_rng(seed)
    return rng.uniform(0, 30, n), rng.uniform(0, 10, n), rng.uniform(0, 100, n)


def run_operator_comparison(n=3000, seed=7):
    """So sánh phép VÀ (min / tích) và phương pháp giải mờ (trọng tâm / MOM)."""
    bs, ss, as_ = _samples(n, seed)
    ref = np.array([reference_label(*t) for t in zip(bs, ss, as_)])
    base = np.array([infer(*t) for t in zip(bs, ss, as_)])
    out = []
    for name, kw in [("min + trọng tâm (đang dùng)", {}),
                     ("tích + trọng tâm", {"tnorm": "prod"}),
                     ("min + trung bình cực đại (MOM)", {"defuzz": "mom"})]:
        y = np.array([infer(*t, **kw) for t in zip(bs, ss, as_)])
        pred = np.array([classify(v) for v in y])
        out.append({"name": name, "agree": float((pred == ref).mean()),
                    "mad": float(np.abs(y - base).mean()), "n_unique": int(len(np.unique(np.round(y, 1))))})
    return out


def run_stability(n=3000, seed=11, noise=0.03):
    """Đo độ ổn định nhãn khi đầu vào bị nhiễu +-3% miền giá trị (mô phỏng sai số đo)."""
    rng = np.random.default_rng(seed)
    bs, ss, as_ = _samples(n, seed)
    rngs = (30, 10, 100)
    nb = np.clip(bs + rng.uniform(-noise, noise, n) * rngs[0], 0, 30)
    ns = np.clip(ss + rng.uniform(-noise, noise, n) * rngs[1], 0, 10)
    na = np.clip(as_ + rng.uniform(-noise, noise, n) * rngs[2], 0, 100)
    f0 = np.array([classify(infer(*t)) for t in zip(bs, ss, as_)])
    f1 = np.array([classify(infer(*t)) for t in zip(nb, ns, na)])
    c0 = np.array([reference_label(*t) for t in zip(bs, ss, as_)])
    c1 = np.array([reference_label(*t) for t in zip(nb, ns, na)])
    y0 = np.array([infer(*t) for t in zip(bs, ss, as_)])
    y1 = np.array([infer(*t) for t in zip(nb, ns, na)])
    return {"n": n, "noise": noise,
            "flip_fuzzy": float((f0 != f1).mean()), "flip_crisp": float((c0 != c1).mean()),
            "mad_y": float(np.abs(y1 - y0).mean()), "max_y": float(np.abs(y1 - y0).max())}


def run_firing_stats(n=3000, seed=5):
    bs, ss, as_ = _samples(n, seed)
    cnt, ys = [], []
    for t in zip(bs, ss, as_):
        y, _, fired, _ = infer(*t, detail=True)
        cnt.append(len(fired))
        ys.append(y)
    cnt = np.array(cnt)
    return {"mean": float(cnt.mean()), "max": int(cnt.max()),
            "dist": {int(k): int((cnt == k).sum()) for k in np.unique(cnt)}}, cnt, np.array(ys)


def plot_block_diagram():
    fig, ax = plt.subplots(figsize=(11, 2.6))
    ax.axis("off")
    boxes = [("Đầu vào rõ\nhành vi, chữ ký,\n% Antivirus", "#e8eef6"),
             ("Mờ hóa\nμ của 9 tập mờ", "#dfeee9"),
             ("Cơ sở luật\n27 luật, VÀ = min", "#fbf0d8"),
             ("Hợp thành + gộp\nmin / max", "#f4e0da"),
             ("Giải mờ\ntrọng tâm", "#e7e1f1"),
             ("Kết luận\nđiểm y* và nhãn", "#e8eef6")]
    w, gap = 1.55, 0.35
    for i, (txt, col) in enumerate(boxes):
        x = i * (w + gap)
        ax.add_patch(plt.Rectangle((x, 0.2), w, 1.2, fc=col, ec="#17365d", lw=1.2))
        ax.text(x + w / 2, 0.8, txt, ha="center", va="center", fontsize=8.5)
        if i < len(boxes) - 1:
            ax.annotate("", xy=(x + w + gap, 0.8), xytext=(x + w, 0.8),
                        arrowprops=dict(arrowstyle="->", color="#17365d", lw=1.4))
    ax.set_xlim(-0.1, len(boxes) * (w + gap))
    ax.set_ylim(0, 1.6)
    fig.tight_layout()
    fig.savefig(f"{OUT}/hinh6_so_do_khoi.png", dpi=170)
    plt.close(fig)


def plot_distribution(ys, cnt):
    fig, axes = plt.subplots(1, 2, figsize=(10, 3.4))
    ax = axes[0]
    ax.hist(ys, bins=np.linspace(0, 100, 41), color="#5b7aa6")
    ax.set(xlabel="Điểm độc hại y*", ylabel="Số mẫu", title="Phân bố đầu ra (3.000 mẫu ngẫu nhiên)")
    ax = axes[1]
    ks = np.unique(cnt)
    ax.bar([str(k) for k in ks], [(cnt == k).sum() for k in ks], color="#7a6fa6")
    ax.set(xlabel="Số luật kích hoạt trong một lần suy diễn", ylabel="Số mẫu", title="Số luật kích hoạt")
    fig.tight_layout()
    fig.savefig(f"{OUT}/hinh7_phan_bo.png", dpi=170)
    plt.close(fig)


def plot_operator_stability(ops, stab):
    fig, axes = plt.subplots(1, 2, figsize=(10, 3.4))
    ax = axes[0]
    names = ["min + TT", "tích + TT", "min + MOM"]
    ax.bar(names, [o["agree"] * 100 for o in ops], color=["#2a6f97", "#8ab17d", "#e9a23b"])
    ax.set(ylabel="Nhất quán với bảng luật (%)", ylim=(80, 100), title="So sánh toán tử / giải mờ")
    for i, o in enumerate(ops):
        ax.text(i, o["agree"] * 100 + 0.4, f'{o["agree"] * 100:.1f}', ha="center", fontsize=8)
    ax = axes[1]
    ax.bar(["Luật rõ\n(ngưỡng cứng)", "Hệ mờ"], [stab["flip_crisp"] * 100, stab["flip_fuzzy"] * 100],
           color=["#c8553d", "#2a9d8f"])
    ax.set(ylabel="Tỷ lệ đổi nhãn khi nhiễu ±3% (%)", title="Độ ổn định khi đầu vào bị nhiễu", ylim=(0, 3.8))
    for i, v in enumerate([stab["flip_crisp"], stab["flip_fuzzy"]]):
        ax.text(i, v * 100 + 0.2, f"{v * 100:.1f}", ha="center", fontsize=8)
    fig.tight_layout()
    fig.savefig(f"{OUT}/hinh8_so_sanh.png", dpi=170)
    plt.close(fig)
```
